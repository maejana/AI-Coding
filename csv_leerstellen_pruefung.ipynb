# Prüfen auf leere Zellen in einer CSV-Datei

Dieses Notebook zeigt, wie eine CSV-Datei mit [pandas](https://pandas.pydata.org/) eingelesen wird und wie man anschließend leere Zellen (fehlende Werte oder leere Zeichenketten) identifiziert.

```python
import pandas as pd
from IPython.display import display

# Pfad zur CSV-Datei anpassen
csv_path = "daten.csv"
```

```python
# CSV einlesen
df = pd.read_csv(csv_path)

# Einen ersten Blick auf die Daten werfen
display(df.head())
```

```python
# Leere Zeichenketten als fehlende Werte behandeln
pruef_df = df.replace(r'^\s*$', pd.NA, regex=True)
fehlende_mask = pruef_df.isna()

if fehlende_mask.any().any():
    gesamt_fehlende = int(fehlende_mask.sum().sum())
    print(f"Es wurden {gesamt_fehlende} leere Zellen gefunden.")
    print("Anzahl leerer Zellen pro Spalte:")
    display(fehlende_mask.sum())
    
    zeilen_mit_luecken = pruef_df[fehlende_mask.any(axis=1)]
    print(f"{len(zeilen_mit_luecken)} Zeilen enthalten mindestens eine leere Zelle.")
    display(zeilen_mit_luecken)
else:
    print("Keine leeren Zellen gefunden.")
```
